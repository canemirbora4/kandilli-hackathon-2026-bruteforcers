```python
import pandas as pd
```

```python
df = pd.read_csv("uzunyil_cleaned.csv")
df.head()
```

```python
# Drop Feb 29 to have consistent 365 days per year (since most years are not leap years)
df = df[df['Date'] != '2012-02-29 00:00:00']

# Transform to long format for consecutive time series
df_long = df.melt(id_vars=['Date'], var_name='Year', value_name='Temperature')

# Create proper date by replacing year in Date with Year column
df_long['Date'] = df_long.apply(lambda row: row['Date'].replace('2012', row['Year']), axis=1)

# Convert Date to datetime
df_long['Date'] = pd.to_datetime(df_long['Date'])

# Sort by Date
df_long = df_long.sort_values('Date').reset_index(drop=True)

# Create df_linear with Date and Temperature
df_linear = df_long[['Date', 'Temperature']]

# Save to CSV
df_linear.to_csv('df_linear.csv', index=False)
print("df_linear saved to df_linear.csv")

# Display the first few rows
df_linear.head()
```
